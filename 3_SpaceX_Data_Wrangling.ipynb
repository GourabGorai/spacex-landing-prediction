# SpaceX Data Wrangling & Feature Engineering
## Objectives
- Perform exploratory data wrangling
- Identify and handle missing payload mass values
- Convert landing outcomes into binary classification target `Class` (1 = Success, 0 = Failure)
- Create One-Hot Encoded features for ML modeling

```
import pandas as pd
import numpy as np

df = pd.read_csv('dataset_part_1.csv')
print('Null count in PayloadMass:', df['PayloadMass'].isnull().sum())
# Impute missing payload mass with mean
df['PayloadMass'].fillna(df['PayloadMass'].mean(), inplace=True)

# Create Landing Outcome Class
landing_outcomes = df['Outcome'].value_counts()
print('Outcome counts:\n', landing_outcomes)

bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])
landing_class = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]
df['Class'] = landing_class
print('Class distribution:\n', df['Class'].value_counts())
df.to_csv('dataset_part_2.csv', index=False)
```
