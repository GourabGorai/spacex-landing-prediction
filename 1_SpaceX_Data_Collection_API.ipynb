# SpaceX API Data Collection & Processing
## Objectives
- Request SpaceX REST API endpoint
- Parse JSON response into structured pandas DataFrame
- Filter Falcon 9 launches & request relational endpoints (cores, payloads, launchpads)

```
import requests
import pandas as pd
import numpy as np
import datetime

# Request launches
url = 'https://api.spacexdata.com/v4/launches/past'
response = requests.get(url)
print('API Status:', response.status_code)
data = response.json()
df = pd.json_normalize(data)
df.head()
```
