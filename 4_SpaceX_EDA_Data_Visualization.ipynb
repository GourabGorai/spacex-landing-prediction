# SpaceX Exploratory Data Analysis with Visualization
## Objectives
- Analyze relationships between Flight Number, Payload Mass, Launch Site, and Orbit
- Plot yearly landing success rate trends

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('dataset_part_2.csv')

# Flight Number vs Launch Site
plt.figure(figsize=(10,5))
sns.scatterplot(y='LaunchSite', x='FlightNumber', hue='Class', data=df, palette={0:'#EF4444', 1:'#10B981'}, s=90)
plt.title('Flight Number vs Launch Site')
plt.show()

# Success Rate by Orbit
orbit_acc = df.groupby('Orbit')['Class'].mean().reset_index()
plt.figure(figsize=(10,5))
sns.barplot(x='Orbit', y='Class', data=orbit_acc)
plt.title('Success Rate by Orbit')
plt.show()
```
