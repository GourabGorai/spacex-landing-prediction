# SpaceX Exploratory Data Analysis (EDA) with Data Visualization

## Project Objectives
In this notebook, we perform Exploratory Data Analysis (EDA) on SpaceX Falcon 9 launch data using `matplotlib` and `seaborn` to extract operational insights across launch sites, payload masses, orbital regimes, and yearly technological maturity.

### 6 Required EDA Visualization Tasks:
1. **Task 1:** Flight Number vs. Launch Site Scatter Chart
2. **Task 2:** Payload Mass vs. Launch Site Scatter Chart
3. **Task 3:** Success Rate vs. Orbit Type Bar Chart
4. **Task 4:** Flight Number vs. Orbit Type Scatter Chart
5. **Task 5:** Payload Mass vs. Orbit Type Scatter Chart
6. **Task 6:** Launch Success Yearly Trend Line Chart

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv('dataset_part_2.csv')
```

## Task 1: Flight Number vs. Launch Site Scatter Chart

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(y="LaunchSite", x="FlightNumber", hue="Class", data=df, palette={0: "#EF4444", 1: "#10B981"}, s=90)
plt.title("1. Flight Number vs. Launch Site", fontsize=12, fontweight='bold')
plt.xlabel("Flight Number", fontsize=10, fontweight='bold')
plt.ylabel("Launch Site", fontsize=10, fontweight='bold')
plt.show()
```

## Task 2: Payload Mass vs. Launch Site Scatter Chart

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(y="LaunchSite", x="PayloadMass", hue="Class", data=df, palette={0: "#EF4444", 1: "#10B981"}, s=90)
plt.title("2. Payload Mass (kg) vs. Launch Site", fontsize=12, fontweight='bold')
plt.xlabel("Payload Mass (kg)", fontsize=10, fontweight='bold')
plt.ylabel("Launch Site", fontsize=10, fontweight='bold')
plt.show()
```

## Task 3: Success Rate vs. Orbit Type Bar Chart

```python
orbit_success = df.groupby('Orbit')['Class'].mean().reset_index().sort_values(by='Class', ascending=False)
plt.figure(figsize=(10, 5))
sns.barplot(x='Orbit', y='Class', data=orbit_success, palette="crest")
plt.title("3. Success Rate by Orbit Type", fontsize=12, fontweight='bold')
plt.xlabel("Orbit Type", fontsize=10, fontweight='bold')
plt.ylabel("Success Rate", fontsize=10, fontweight='bold')
plt.show()
```

## Task 4: Flight Number vs. Orbit Type Scatter Chart

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(y="Orbit", x="FlightNumber", hue="Class", data=df, palette={0: "#EF4444", 1: "#10B981"}, s=90)
plt.title("4. Flight Number vs. Orbit Type", fontsize=12, fontweight='bold')
plt.xlabel("Flight Number", fontsize=10, fontweight='bold')
plt.ylabel("Orbit Type", fontsize=10, fontweight='bold')
plt.show()
```

## Task 5: Payload Mass vs. Orbit Type Scatter Chart

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(y="Orbit", x="PayloadMass", hue="Class", data=df, palette={0: "#EF4444", 1: "#10B981"}, s=90)
plt.title("5. Payload Mass (kg) vs. Orbit Type", fontsize=12, fontweight='bold')
plt.xlabel("Payload Mass (kg)", fontsize=10, fontweight='bold')
plt.ylabel("Orbit Type", fontsize=10, fontweight='bold')
plt.show()
```

## Task 6: Launch Success Yearly Trend Line Chart

```python
df['Year'] = pd.to_datetime(df['Date']).dt.year
yearly_success = df.groupby('Year')['Class'].mean().reset_index()
plt.figure(figsize=(10, 5))
plt.plot(yearly_success['Year'], yearly_success['Class'], marker='o', color='#0284C7', linewidth=2.5)
plt.title("6. Launch Success Yearly Trend Line (2010 - 2020)", fontsize=12, fontweight='bold')
plt.xlabel("Year", fontsize=10, fontweight='bold')
plt.ylabel("Average Success Rate", fontsize=10, fontweight='bold')
plt.show()
```
