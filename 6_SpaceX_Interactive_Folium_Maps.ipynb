# SpaceX Launch Site Proximity Analysis with Folium
## Objectives
- Mark launch sites on interactive Folium map
- Add Marker Clusters for landing success/failure outcomes
- Calculate geodesic distance to coastlines, railways, highways, and cities

```
import folium
from folium.plugins import MarkerCluster
import pandas as pd

df_geo = pd.read_csv('spacex_launch_geo.csv')
site_map = folium.Map(location=[28.562302, -80.577356], zoom_start=5)
marker_cluster = MarkerCluster().add_to(site_map)

for index, row in df_geo.iterrows():
    marker = folium.Marker(
        location=[row['Lat'], row['Long']],
        icon=folium.Icon(color='green' if row['class']==1 else 'red'),
        popup=f"{row['Launch Site']} (Class: {row['class']})"
    )
    marker_cluster.add_child(marker)

site_map.save('spacex_launch_site_map.html')
print('Interactive Folium Map saved as spacex_launch_site_map.html')
```
