# SpaceX Falcon 9 Launch Records Web Scraping
## Objectives
- Scrape Wikipedia Falcon 9 launch table using BeautifulSoup
- Extract Launch Number, Date, Booster Version, Launch Site, Payload, Mass, Orbit, Outcome

```
import bs4
from bs4 import BeautifulSoup
import requests
import pandas as pd

static_url = 'https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches'
response = requests.get(static_url)
soup = BeautifulSoup(response.text, 'html.parser')
print('Page Title:', soup.title.string)
```
